```python
#Removing duplicate rows
df.drop_duplicates(inplace = True)
print("Shape after removing duplicates: ",df.shape)
```

```python
#Handle missing values
print("Missing values before cleaning :\n")
print(df.isnull().sum())
df=df.dropna(subset=['Smiles','Standard Type','Standard Value','Standard Units']).reset_index(drop=True)
print("Shape after removing required missing values: ",df.shape)
```

```python
print(df['Standard Type'].value_counts())
```

```python
df_ic50=df[df['Standard Type'].str.upper().str.strip()=="IC50"].copy()
print(df_ic50.shape)
```

```python
df=df[df['Standard Value']!=0].copy() #removing the rows with 0 Standard Value
df.reset_index(drop=True,inplace=True)
print("Minimum IC50:", df["Standard Value"].min())
print("Maximum IC50:", df["Standard Value"].max())
print("Median IC50:", df["Standard Value"].median())
print("Mean IC50:",df['Standard Value'].mean())
```

```python
#No Outliers exist
df['logIC50']=np.log10(df['Standard Value'])
Q1=df['logIC50'].quantile(0.25)
Q3=df['logIC50'].quantile(0.75)
IQR=Q3-Q1
l=Q1-1.5*IQR
u=Q3+1.5*IQR
before=len(df)
df=df[(df['logIC50']>=l) & (df['logIC50']<=u)].copy()
after=len(df)
print("Rows before outlier removal:", before)
print("Rows after outlier removal:", after)
print("Outliers removed:", before - after)
```

```python
def convert_to_nM(row):
    value = row["Standard Value"]
    unit = row["Standard Units"]

    if unit == "nm":
        return value
    elif unit == "um":
        return value * 1000
    elif unit == "mm":
        return value * 1_000_000
    else:
        return np.nan

df["Standard Value"] = df.apply(
    convert_to_nM,
    axis=1
)

print(df['Standard Units'].value_counts())
```

```python
#Creating Activity Colomn
df['Activity']=np.where(df['Standard Value']<=1000,1,0)
print(df['Activity'].value_counts())
```
